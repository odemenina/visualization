```python
import plotly.graph_objects as go
```

```python
x1 = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanju'])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
graph = go.Figure()
```

```python
graph.add_trace(go.Bar(x = x1, y = y1))
```

```python
graph.update_layout (title = 'Pirate Scores', 
                     xaxis_title = 'Pirates', 
                    yaxis_title = 'Scores')
```

```python
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1,
                           y=y1,
                           mode='markers',                                 #we want points for scatter plot
                           marker = dict(
                           size = 15,                                  #point size
                           color = 'hotpink',                          # point colour
                           opacity = 1,                                # point transparency/alpha
                           line = dict(width = 5, color = 'purple')    # plont outline
                           )))
graph.update_layout(
    title = 'Interactive Pirate Plot',
    xaxis_title = 'Pirates',
    yaxis_title = 'Scores',
    width =500,     
    height = 500)


```

```python
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
                 on_bad_lines = 'skip')
df
```

```python
#join all our text form each row from our quote column nto a string
text =  " ".join(each for each in df.quote)

#generate our wordcoud image

wordcloud = WordCloud(background_color = "white", colormap = 'inferno').generate(text)

#use matplotlib syntax to put image in a figure
fig, ax =  plt.subplots(figsize = (7, 3))
ax.imshow(wordcloud,                   #remeber 'imshow'form when we added pictures to our matplotlib axes
          interpolation = 'bilinear')   #this line helps smooth our image

ax.axis("off")
plt.show()
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
A = set(['apple', 'banana', 'watermelon'])
B = set (['apple','pumpkin', 'blueberry', 'key limme'])
diagram = venn2([A,B],
                set_labels = ('Fruits', 'Pires'),
                set_colors = ('blue', 'red'),
                alpha = 0.5)

diagram.get_label_by_id("10").set_text("\n".join(A-B)) #left
diagram.get_label_by_id("11").set_text("\n".join(A&B)) #middle
diagram.get_label_by_id("01").set_text("\n".join(B-A)) #right
```
